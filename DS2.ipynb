```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5,6,7]
y = [0,9,9,8,7,6,5]

plt.plot(x,y)
plt.show()
```

```python
months =["jan","feb","mar","apr","may","jun","jul"]
sales =[100,20,30,55,33,88,66]

plt.plot(months,sales,marker="o",linestyle="--")
plt.title("monthly sales")
plt.xlabel("months")
plt.ylabel("sales")
```

```python
months =["jan","feb","mar","apr","may","jun","jul"]
sales1 =[100,20,30,55,33,88,66]
sales2 =[110,30,40,65,73,48,16]

plt.plot(months,sales1,label="product a")
plt.plot(months,sales2,label="product b")

plt.title("product sales comparison")
plt.xlabel("Months")
plt.ylabel("sales")

plt.legend()
plt.show()
```

```python
months =["jan","feb","mar","apr","may","jun","jul"]
sales =[100,20,30,55,33,88,66]

plt.barh(months,sales)
plt.title("monthly sales")
plt.xlabel("months")
plt.ylabel("sales")
```

```python
months =["jan","feb","mar","apr","may","jun","jul"]
sales =[100,20,30,55,33,88,66]

plt.scatter(months,sales)
plt.title("monthly sales")
plt.xlabel("months")
plt.ylabel("sales")
plt.show()
```

```python
subjects=["python","java","C","go"]
students=[20,40,10,30]
plt.pie(students,labels=subjects,autopct='%1.1f%%')
plt.title("count")
plt.show
```

```python
marks = [45,55,65,75,85,25,250]

plt.boxplot(marks)

plt.title("students mark")
plt.ylabel("Marks")
```

```python
x=[1,2,3,4,5]
y1=[10,20,30,40,50]
y2=[50,40,30,20,10]

plt.subplot(1,2,1)
plt.plot(x,y1)
plt.title("increasing")

plt.subplot(1,2,2)
plt.plot(x,y2)
plt.title("decreasing")
```
