1.Program to multiply two given arrays of same size element-by-element.

```python
import numpy as np
arr1=np.array([1,2,3,4])
arr2=np.array([3,4,2,5])
mul=np.multiply(arr1,arr2)
print('Array multiplication:',mul)
```

2. Program to create an element-wise comparison (greater, greater equal, lesser).

```python
import numpy as np
arr1=np.array([1,6,3,4,])
arr2=np.array([2,4,3,5,])
print('Greater:', np.greater(arr1,arr2))
print('Greater or equal:',np.greater_equal(arr1,arr2))
print('Lesser:',np.less(arr1,arr2))
print('Lesser or equal:',np.less_equal(arr1,arr2))
```
